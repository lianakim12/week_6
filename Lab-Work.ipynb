```python
# import libraries

import pandas as pd
import numpy as np
import sklearn as sk

# load data

salary_data = pd.read_csv("/workspaces/week_6/2025_salaries.csv", header=1, encoding="latin-1")
stats = pd.read_csv("/workspaces/week_6/nba_2025.txt", sep=",", encoding="latin-1")
```

```python
# merge the dataframes based on Player

merged_data = pd.merge(salary_data, stats, on="Player")

# check for duplicate Player rows

duplicates = merged_data[merged_data.duplicated(subset="Player", keep=False)]
duplicates.head()
```

```python
# sort so that rows with highest G come first

merged_data = merged_data.sort_values(by='G', ascending=False)

# drop duplicate player rows that are not the highest G

merged_cleaned = merged_data.drop_duplicates(subset='Player', keep='first').reset_index(drop=True)
merged_cleaned.head()
```

```python
# check for variables with null values
# merged_cleaned.isnull().sum()
```

```python
# keep columns that are likely the most related to salary

columns_to_keep = ['Player', '2025-26', 'Rk', 'Age', 'G', '3P%', '2P%', 'eFG%', 'FT%', 'TRB', 
                   'PTS', 'Trp-Dbl']
X = merged_cleaned[columns_to_keep]

# drop null values

X = X.dropna()
X.head()
```

```python
# remove $ and , in 2025-26 and convert into float

X["2025-26"] = X["2025-26"].str.replace("[$,]", "", regex=True).astype(float)
X.head()
```

```python
# scale only numeric columns

from sklearn.preprocessing import StandardScaler

X_scaled = X.copy()
scaler = StandardScaler()
X_scaled.iloc[:, 1:] = scaler.fit_transform(X.iloc[:, 1:])
X_scaled.head()
```

```python
# prep data for model by dropping Player and salary

features = X_scaled.drop(columns=["Player", "2025-26"])
```

```python
# KMeans clustering with best guess for k

from sklearn.cluster import KMeans
kmeans_obj = KMeans(n_clusters=5, random_state=42).fit(features)
```

```python
# examine clustering results

print(f"Cluster centers: {kmeans_obj.cluster_centers_}")
print(f"Cluster labels: {kmeans_obj.labels_}")
print(f"Inertia: {kmeans_obj.inertia_}")
```

```python
# visualize clusters in 3D

import plotly.express as px

fig = px.scatter_3d(
    X_scaled, x="eFG%", y="PTS", z="G", # 3 chosen variables
    color="2025-26", # color indicates salary
    symbol=kmeans_obj.labels_, # shape indicates cluster
    hover_data=["Player"]) # check which point is which Player
fig.show(renderer="browser")
```

```python
# calculate variance explained

# first calculate total sum of squares (TSS)
total_sum_squares = np.sum((features - np.mean(features))**2)
total = np.sum(total_sum_squares)

# then calculate between-cluster sum of squares (BSS)
between_SSE = (total - kmeans_obj.inertia_)

# finally calculate variance explained (BSS / TSS )
var_explained = between_SSE / total
print(f"Variance Explained: {var_explained:.4f} or {var_explained*100:.2f}%")
```

```python
# calculate silhouette score

from sklearn.metrics import silhouette_score
ss = silhouette_score(features, kmeans_obj.labels_)
print(f"Silhouette score: {ss}")
```

```python
# elbow curve: first calculate within-cluster sum of squares (WCSS) for different k

wcss = []
for i in range(1, 11):
    kmeans_obj = KMeans(n_clusters=i, random_state=42).fit(features)
    wcss.append(kmeans_obj.inertia_)
```

```python
# elbow curve: plot using the WCSS for different k

elbow_data_Dem = pd.DataFrame({"k": range(1, 11), "wcss": wcss})
fig = px.line(elbow_data_Dem, x="k", y="wcss", title="Elbow Method using WCSS")
fig.show()
```

```python
# elbow curve: first calculate within-cluster sum of squares (WCSS) for different k

sss = []
for i in range(2, 11): # silhouette score cannot run for k=1
    kmeans_obj = KMeans(n_clusters=i, random_state=42).fit(features)
    ss = silhouette_score(features, kmeans_obj.labels_)
    sss.append(ss)
```

```python
# elbow curve: plot using the silhouette score for different k

elbow_data_Dem = pd.DataFrame({"k": range(2, 11), "ss": sss})
fig = px.line(elbow_data_Dem, x="k", y="ss", title="Elbow Method using Silhouette Score")
fig.show()
```

```python
# WCSS: elbow bends at k=2
# silhouette score: highest at k=2

# based on the two elbow plots, best k = 2

kmeans_obj_2 = KMeans(n_clusters=2, random_state=42).fit(features)
```

```python
# visualize clusters in 3D

fig = px.scatter_3d(
    X_scaled, x="eFG%", y="PTS", z="G", # 3 chosen variables
    color="2025-26", # color indicates salary
    symbol=kmeans_obj.labels_, # shape indicates cluster
    hover_data=["Player"]) # check which point is which Player
fig.show(renderer="browser")
```

```python
# final model evaluation: calculate variance explained

# first calculate total sum of squares (TSS)
total_sum_squares = np.sum((features - np.mean(features))**2)
total = np.sum(total_sum_squares)

# then calculate between-cluster sum of squares (BSS)
between_SSE = (total - kmeans_obj_2.inertia_)

# finally calculate variance explained (BSS / TSS )
var_explained = between_SSE / total
print(f"Variance Explained: {var_explained:.4f} or {var_explained*100:.2f}%")
```

```python
# final model evaluation: calculate silhouette score

from sklearn.metrics import silhouette_score
ss = silhouette_score(features, kmeans_obj_2.labels_)
print(f"Silhouette score: {ss}")
```

What sort of players do we want?

We want players that perform well but are underpaid. In other words, we want players that are shown as circles ( = clustered as the relatively better performers) with a dark color ( = getting relatively low salary) in the 3D visualization. We do NOT want players that are shown as squares ( = clustered as the relatively worse performers) with a light color ( = getting relatively high salary). We also want to check the individual players' stats to ensure they are the player we want before making any final decisions.
